#### **Q1 — Positive, Negative or Zero**

```python
number = float(input("Enter a Number you want to check that is +ve,-ve,0: "))
if number ==0:
    print(f"{number} is Zero")
elif number > 0:
    print(f"{number} is positive")
elif number < 0:
    print(f"{number} is negative")
else:
    print("Not identified")
```

#### **Q2 — Even or Odd**

```python
number = int(input("Enter Number: "))
if number % 2 == 0:
    print(f"{number} is even")
elif number % 2 != 0:
    print(f"{number} is odd")
else:
    print("Dont Know!")
```

#### **Q3 — Largest of Two Numbers**

```python
num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
if num1 <  num2:
    print(f"{num2} is greater than {num1}")
else:
    print(f"{num1} is greater than {num2}")
```

#### **Q4 — Sum from 1 to N**

```python
n = int(input("Enter a number: "))
s = 0
for i in range(1,n+1):
    s = s+ i
print(s)
```

#### **Q5 — Count Even Numbers**

```python
numbers = [2, 7, 10, 13, 18, 21]
c = 0
for i in numbers:
    if i % 2 ==0:
        c += 1
print(c)
```

#### **Q6 — Reverse a String**

```python
s = "python"
rt = "".join(reversed(s))
print(rt)
```

#### **Q7 — Count Vowels**

```python
s = "machine learning"
c = 0
v = ["a", "e", "i", "o", "u"]
for i in s:
    if i in v:
        c += 1
print(c)
```

#### **Q8 — Find Maximum in a List**

```python
numbers = [12, 45, 7, 89, 23]
max = 0
for i in numbers:
    if i > max:
        max = i
print(max)
```

#### **Q9 — Count Frequency ⭐**

```python
numbers = [1, 2, 2, 3, 1, 2, 4, 3]
freq = {}
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
freq
```

#### **Q10 — FizzBuzz 🧠**

```python
# divisible by 3 --> Fizz
# divisible by 5 --> Buzz
# divisible by 3 and 5 --> FizzBuzz
# else --> number

num = int(input("Enter a number: "))
for i in range(1,num+1):
    if (i % 3 == 0) and (i %5 == 0):
        print("Fizz Buzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz") 
    else:
        print(i)
```
