#### **Q9 — Create a Function for Statistics**

###### **Write a function:**

###### **calculate_stats(numbers)**

###### **It should return:**

###### **minimum**
###### **maximum**
###### **average**
###### **count**

###### **For:**

###### **numbers = [10, 20, 30, 40, 50]**

```python
def calculate_stats(numbers):
    minimum_num = min(numbers)
    maximum_num = max(numbers)
    average = sum(numbers)/len(numbers)
    count = len(numbers)
    return minimum_num, maximum_num, average, count
a = calculate_stats([10, 20, 30, 40, 50])
print(a)
```

#### **Q10 — Simple ML Data Preprocessing 🧠🤖**

###### **Imagine these are student scores:**

###### **scores = [45, 78, 32, 90, 66, 55, 88, 40]**

###### **Create a program that:**

###### **Finds the average score.**
###### **Creates a new list containing only scores greater than or equal to the average.**
###### **Creates another list containing scores below the average.**
###### **Counts how many students are above/equal to average.**
###### **Counts how many are below average.**

```python
scores = [45, 78, 32, 90, 66, 55, 88, 40]
def simple_ml(scores):
    avg_score = sum(scores)/len(scores)
    egavg_score = [i for i in scores if i >= avg_score]
    elavg_score = [i for i in scores if i < avg_score]
    ae_avg = len(egavg_score)
    ab_avg = len(elavg_score)
    return f"""
Average = {avg_score},
Greater than or equal to the average = {egavg_score},
Below the average = {elavg_score},
Count above/equal to average = {ae_avg},
Count below avg = {ab_avg}

"""

print(simple_ml([45, 78, 32, 90, 66, 55, 88, 40]))
```

text = "python machine learning"

Produce:

nohtyp enihcam gninrael

Rules:

You may use split()
Don't use [::-1]
Don't use reversed()

```python
text = "python machine learning"

words = text.split()
result = []

for word in words:
    reversed_word = ""

    for i in range(len(word) - 1, -1, -1):
        reversed_word += word[i]

    result.append(reversed_word)

print(" ".join(result))
```

Q9 — Word Frequency Function 🤖

Create:

def word_frequency(text):
    ...

For:

text = "Python is great and Python is powerful"

Return a dictionary containing the frequency of each word.

Expected conceptually:

{
    "python": 2,
    "is": 2,
    "great": 1,
    "and": 1,
    "powerful": 1
}

```python
def word_frequency(text):
    list_text = text.split()
    freq = {}
    for i in list_text:
        if i in freq:
            freq[i] += 1
        else:
            freq[i] = 1
    return freq
text = "Python is great and Python is powerful"
print(word_frequency(text))
```
