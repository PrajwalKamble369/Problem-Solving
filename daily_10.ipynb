#### **Q1 — Positive, Negative or Zero**

```python
number = float(input("Enter a Number you want to check that is +ve,-ve,0: "))
if number ==0:
    print(f"{number} is Zero")
elif number > 0:
    print(f"{number} is positive")
elif number < 0:
    print(f"{number} is negative")
else:
    print("Not identified")
```

#### **Q2 — Even or Odd**

```python
number = int(input("Enter Number: "))
if number % 2 == 0:
    print(f"{number} is even")
elif number % 2 != 0:
    print(f"{number} is odd")
else:
    print("Dont Know!")
```

#### **Q3 — Largest of Two Numbers**

#### **Q4 — Sum from 1 to N**

```python
n = int(input("Enter a number: "))
s = 0
for i in range(1,n+1):
    s = s+ i
print(s)
```

#### **Q5 — Count Even Numbers**

```python
numbers = [2, 7, 10, 13, 18, 21]
c = 0
for i in numbers:
    if i % 2 ==0:
        c += 1
print(c)
```

#### **Q6 — Reverse a String**

```python
s = "python"
rt = "".join(reversed(s))
print(rt)
```

#### **Q7 — Count Vowels**

```python
s = "machine learning"
c = 0
v = ["a", "e", "i", "o", "u"]
for i in s:
    if i in v:
        c += 1
print(c)
```

#### **Q8 — Find Maximum in a List**

```python
numbers = [12, 45, 7, 89, 23]
max = 0
for i in numbers:
    if i > max:
        max = i
print(max)
```

#### **Q9 — Count Frequency ⭐**

```python
numbers = [1, 2, 2, 3, 1, 2, 4, 3]
freq = {}
for i in numbers:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
freq
```

#### **Q10 — FizzBuzz 🧠**

```python
# divisible by 3 --> Fizz
# divisible by 5 --> Buzz
# divisible by 3 and 5 --> FizzBuzz
# else --> number

num = int(input("Enter a number: "))
for i in range(1,num+1):
    if (i % 3 == 0) and (i %5 == 0):
        print("Fizz Buzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz") 
    else:
        print(i)
```

#### **Q1 — Count Positive, Negative & Zero**

##### **Given a list of numbers, count how many are:**

###### **positive**
###### **negative**
###### **zero**

###### **Example:**

###### **numbers = [10, -5, 0, 7, -2, 0, 8]**

```python
numbers = [10, -5, 0, 7, -2, 0, 8]
positive = 0
negative = 0
zero = 0
for i in numbers:
    if i >0:
        positive += 1
    elif i < 0:
        negative += 1
    else:
        zero +=1
print(positive,negative,zero)
```

#### **Q2 — Sum of Even and Odd Numbers**

###### **Given a list, calculate separately:**

###### **sum of even numbers**
###### **sum of odd numbers**

###### **Example:**

###### **numbers = [1, 2, 3, 4, 5, 6]**

```python
numbers = [1, 2, 3, 4, 5, 6]
sum_even = 0
sum_odd = 0
for i in numbers:
    if i % 2 == 0:
        sum_even += i
    else:
        sum_odd +=i
print(sum_even,sum_odd)
```

#### **Q3 — Find the Smallest Number**

###### **Find the smallest number in a list without using min().**

###### **Test your code with:**

###### **numbers = [12, 5, 89, 3, 17, 1, 45]**

```python
numbers = [12, 5, 89, 3, 17, 1, 45]
min_num = numbers[0]
for i in numbers:
    if i < min_num:
        min_num = i
print(min_num)
```

#### **Q4 — Reverse a String Without reversed() or [::-1]**

###### **Given:**

###### **text = "developer"**

```python
text = "developer"
r_t = ""
for i in range(len(text)-1,-1,-1):
    r_t += text[i]
print(r_t)
```

#### **Q5 — Palindrome Checker**

###### **Write a program that checks whether a string is a palindrome.**

###### **Examples:**

###### **madam → Palindrome**
###### **python → Not Palindrome**
###### **level → Palindrome**

```python
text = input("enter text: ").lower()
rt = text[::-1]
if text == rt:
    print("Palindrome")
else:
    print("Not Palindrome")
```
